```python
import matplotlib.pyplot as plt
import numpy as np

layer_sizes = [3, 5, 4, 2]
layer_names = [
    "Capa de Entrada\n(Input Layer)",
    "Capa Oculta 1\n(Hidden Layer 1)",
    "Capa Oculta 2\n(Hidden Layer 2)",
    "Capa de Salida\n(Output Layer)"
]
colors = ['#3498db', '#2ecc71', '#e67e22', '#e74c3c']

fig, ax = plt.subplots(figsize=(12, 7))

h_spacing = 2.0
neurons = []

for i, size in enumerate(layer_sizes):
    layer_neurons = []
    y_coords = np.arange(size) - (size - 1) / 2.0
    for y in y_coords:
        layer_neurons.append((i * h_spacing, y))
    neurons.append(layer_neurons)


for i in range(len(layer_sizes) - 1):
    for n1 in neurons[i]:
        for n2 in neurons[i+1]:
            ax.plot([n1[0], n2[0]], [n1[1], n2[1]], color='#bdc3c7', alpha=0.5, zorder=1, linewidth=1)


for i, layer_neurons in enumerate(neurons):
    for n in layer_neurons:
        circle = plt.Circle(n, 0.28, color=colors[i], ec='black', linewidth=1.5, zorder=2)
        ax.add_patch(circle)


max_height = max(layer_sizes) / 2.0
for i, name in enumerate(layer_names):
    ax.text(i * h_spacing, max_height + 1.2, name,
            ha='center', va='center', fontsize=11, fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.5", fc="#f8f9fa", ec="#bdc3c7", lw=1))


ax.set_xlim(-1, len(layer_sizes) * h_spacing - 1)
ax.set_ylim(-max_height - 1.8, max_height + 2.2)
ax.axis('off')
plt.title("Diseño de Red Neuronal Profunda (DNN)", fontsize=15, fontweight='bold', pad=25)
plt.tight_layout()
plt.show()
```
