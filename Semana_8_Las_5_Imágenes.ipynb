```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files

# Buscamos si el archivo existe con su nombre original o con sufijos numéricos
posibles_nombres = ['Mango Desenfocado (2).jpg', 'Mango Desenfocado (1).jpg', 'Mango Desenfocado.jpg']
nombre_real = next((f for f in posibles_nombres if os.path.exists(f)), None)

# Si no existe en el entorno, pedimos subirlo directamente
if nombre_real is None:
    print("No se encontró el archivo desenfocado. Por favor, selecciona y sube 'Mango Desenfocado.jpg':")
    subido = files.upload()
    if subido:
        nombre_real = list(subido.keys())[0]

if nombre_real is not None:
    imagen = cv2.imread(nombre_real)

    # Definimos el kernel de enfoque
    kernel_enfoque = np.array([[ 0, -1,  0],
                               [-1,  5, -1],
                               [ 0, -1,  0]])

    # Aplicamos el filtro para mejorar la nitidez
    imagen_enfocada = cv2.filter2D(imagen, -1, kernel_enfoque)

    # Guardamos el resultado
    cv2.imwrite('imagen_enfocada.jpg', imagen_enfocada)

    # Mostramos la comparación
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.title("Original")
    plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title("Enfocada / Nitidez mejorada")
    plt.imshow(cv2.cvtColor(imagen_enfocada, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
else:
    print("No se pudo realizar el proceso porque no se proporcionó el archivo requerido.")
```

```python
from PIL import Image, ImageFilter

# 1. Cargar la imagen
img = Image.open('Mango Desenfocado.jpg')

# 2. Aplicar Máscara de Enfoque
# radius: qué tan gruesos son los bordes a realzar
# percent: la intensidad del enfoque (ej. 150 a 200 suele funcionar bien)
# threshold: diferencia mínima de contraste para evitar amplificar el ruido
img_sharp = img.filter(ImageFilter.UnsharpMask(radius=2, percent=150, threshold=3))

# 3. Guardar el resultado
img_sharp.save('imagen_unsharp.jpg')
print("¡Imagen enfocada con Máscara de Enfoque!")
```

```python
import cv2
import numpy as np

# 1. Cargar imagen
img = cv2.imread('Mango Desenfocado.jpg')

# 2. Aplicar filtro bilateral para reducir ruido sin perder bordes
# (d=9, sigmaColor=75, sigmaSpace=75 son valores estándar de prueba)
suavizada = cv2.bilateralFilter(img, 9, 75, 75)

# 3. Aplicar un kernel de enfoque más potente sobre la imagen limpia
kernel_fuerte = np.array([[-1, -1, -1],
                          [-1,  9, -1],
                          [-1, -1, -1]])

imagen_final = cv2.filter2D(suavizada, -1, kernel_fuerte)

# 4. Guardar resultado
cv2.imwrite('imagen_mejorada_bilateral.jpg', imagen_final)
print("¡Proceso completado!")
```

```python
import cv2
import numpy as np

def calcular_metricas(ruta_imagen):
    img = cv2.imread(ruta_imagen)
    if img is None:
        return None

    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    brillo = np.mean(gris)

    contraste = np.std(gris)

    nitidez = cv2.Laplacian(gris, cv2.CV_64F).var()

    alto, ancho = gris.shape

    return {
        'brillo': round(brillo, 2),
        'contraste': round(contraste, 2),
        'nitidez': round(nitidez, 2),
        'resolucion': f"{ancho}x{alto} px"
    }


m_antes = calcular_metricas('Mango referencia.jpg')
m_despues = calcular_metricas('imagen_mejorada.jpg')

if m_antes and m_despues:
    print("--- RESULTADOS PARA LA TABLA ---")
    print(f"Brillo     -> Antes: {m_antes['brillo']} | Después: {m_despues['brillo']}")
    print(f"Contraste  -> Antes: {m_antes['contraste']} | Después: {m_despues['contraste']}")
    print(f"Nitidez    -> Antes: {m_antes['nitidez']} | Después: {m_despues['nitidez']}")
    print(f"Resolución -> Antes: {m_antes['resolucion']} | Después: {m_despues['resolucion']}")
else:
    print("Verifica los nombres y rutas de los archivos de imagen.")
```
